# Preprocessing

```python
import pyodbc
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df_test = pd.read_csv('../../../decoded_data/PAQ/FactTest.csv')
df_question = pd.read_csv('../../../decoded_data/PAQ/FactQuestionPAQ.csv')
```

```python
df_test.drop(columns=['VersionNumber', 'Test'], inplace=True)
df_test.head()
```

```python
df_question.drop(columns=['InstanceID', 'Test'], inplace=True)
df_question.head()
```

```python
df = df_question.merge(df_test, on='TestKey')
df.head()
```

### CreatedDate

```python
df["CreatedDateKey"] = pd.to_datetime(df.CreatedDateKey, format="%Y%m%d")
df.head()
```

```python
df.info()
```

```python
df['Year'] = pd.DatetimeIndex(df['CreatedDateKey']).year
df['Month'] = pd.DatetimeIndex(df['CreatedDateKey']).month
df['Day'] = pd.DatetimeIndex(df['CreatedDateKey']).dayofweek
df.drop(columns=['CreatedDateKey'], inplace=True)
df.info()
```

```python
df.head()
```

## Cleaning

```python
df.info()
```

```python
df.dropna(inplace=True)
df.info()
```

### Make numerical

```python
df.info()
df.head()
```

```python
only_int = df.AnswerVal.apply(lambda x: x % 1 == 0).all()
if only_int:
    print('In the column AnswerVal none of the values have a value that needs to be stored as a float, therefor we can change the datatype to integer.')
    df['AnswerVal'] = df['AnswerVal'].astype(int)
df.info()
```

### Foutieve waarden

```python
is_valid = df['AnswerVal'].isin([1, 2, 3, 4, 5]).all()

if not is_valid: 
    print("Some values in the AnswerVal column do not fall in the permitted range.")
    df = df[df['AnswerVal'].isin([1, 2, 3, 4, 5])]
else: 
    print("All values in the AnswerVal column fall in the permitted range.")
```

## Preprocessing per question

```python
df.columns
```

#### Statement specific

```python
df_statement_specific = df[['QuestionKey', 'LeftStatement', 'RightStatement', 'AnswerVal',
       'TestKey', 'CandidateKey', 'Year', 'Month', 'Day']]
df_statement_specific.head()
```

```python
df_statement_specific['StatementPair'] = df_statement_specific.LeftStatement + df_statement_specific.RightStatement
```

```python
df_statement_specific.head()
```

```python
le = LabelEncoder()
df_statement_specific['StatementPair'] = le.fit_transform(df_statement_specific['StatementPair'])
```

```python
df_statement_specific.drop(columns=['LeftStatement', 'RightStatement'], inplace=True)
df_statement_specific.head()
```

#### Statement type specific 

```python
df.columns
```

```python
df_type = df[['QuestionKey', 'LeftStatement', 'RightStatement', 'AnswerVal',
       'TestKey', 'CandidateKey', 'Year', 'Month', 'Day']]
df_type.head()
```

```python
df_type['LeftStatement_type'] = df_type['LeftStatement'].str.split('_').str[0]
df_type['RightStatement_type'] = df_type['RightStatement'].str.split('_').str[0]
df_type.drop(columns=['RightStatement', 'LeftStatement'], inplace=True)
df_type.head()
```

```python
df_type['StatementTypePair'] = df_type.LeftStatement_type + '_' + df_type.RightStatement_type
```

```python
df_type.head()
```

```python
df_type.drop(columns=['LeftStatement_type', 'RightStatement_type'], inplace=True)
df_type.head()
```
