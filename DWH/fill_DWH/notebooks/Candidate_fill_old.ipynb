imports

```python
import pandas as pd
from dep2pyodbc import dep2connection
```

Connection to database

```python
DATABASE = 'CRH'
DB_DWH = 'CRH_DWH'

channel_dwh = dep2connection(DATABASE, True)
channel_dwh_lisa = dep2connection(DB_DWH)
cursor_dwh_lisa = channel_dwh_lisa.cursor()
```

Get data from candidate table in CRH en put it in a dataframe

```python
df_candidate = pd.read_sql("SELECT * FROM Candidate", channel_dwh)

df_candidate.info()
```

Use only columns that are needed in the DWH from the dataframe

```python
df_candidate_dwh = df_candidate[["ID", "InstanceID", "OrganizationGUID", "LanguageGUID", "GenderChoice", "Gender", "Qualification"]]

df_candidate_dwh.head(20)
```

Add CandidateKey by combining ID and Instance into 1 code: {ID}{InstanceID}

```python
df_candidate_dwh["CandidateKey"] = df_candidate_dwh["ID"].astype(str) + df_candidate_dwh["InstanceID"].astype(str)
df_candidate_dwh["CandidateKey"].astype(int)

df_candidate_dwh.head(20)
```

```python
df_candidate_dwh["CandidateKey"].describe()
```

Make sure columns are in the same order as the ones in the database

```python
df_candidate_dwh = df_candidate_dwh[["CandidateKey", "ID", "InstanceID", "OrganizationGUID", "LanguageGUID", "GenderChoice", "Gender", "Qualification"]]

df_candidate_dwh.head()
```

```python
df_candidate_dwh.rename(columns={'OrganizationGUID': 'Organisation',
                                 'LanguageGUID': 'ChosenLanguage', 
                                 'GenderChoice': 'ChosenGender',}, 
    inplace=True)
df_candidate_dwh.head()
```

Write the dataframe to DimCandidate in the DWH

```python
# df_candidate_dwh.to_sql("DimCandidate", channel_dwh, index=False, if_exists="append")
```

```python
# df_candidate_dwh.to_csv("../csv/DWH structure/DimCandidate.csv", index=False)
```

```python
for index, row in df_candidate_dwh.iterrows():
    row = list(row)
    query = """INSERT INTO DimCandidate(CandidateKey, ID, InstanceID, Organisation, ChosenLanguage, ChosenGender, Gender, Qualification) VALUES (?,?,?,?,?,?,?,?)"""
    cursor_dwh_lisa.execute(query, row)
channel_dwh_lisa.commit()
```
